# Import

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
sns.set_theme()
```

```python
X = np.array([
    [1,4, 1.5, 3.8],
    [3,6, 1.0, 3.1],
    [4,7, 0.8, 3.9],
    [6,8, 0.4, 3.4],
    [7,9, 0.2, 2.8]
])

y = np.array([45, 60, 72, 85, 95])

df = pd.DataFrame(
    X,
    columns=['Year_of_exp', 'PS_Skill', 'Pause_time', 'GPA']
)

df['Salary'] = y
df
```

```python
def multiple_linear_regression(X, y, w, b):
    m = X.shape[0]
    prediction_list = np.zeros((m,))
    for i in range(m):
        prediction_list[i] = np.dot(w, X[i]) + b
    return prediction_list
```

```python
W_init = np.ones((X.shape[1],))
b_init = 0
prediction_list = multiple_linear_regression(X, y, W_init, b_init)
```

## Mean Squared Error

```python
value = prediction_list - y
squared_errors = value ** 2
sum_squared_errors = np.sum(squared_errors)
mean_squared_error = sum_squared_errors / X.shape[0]

print(f"Mean Squared Error: {mean_squared_error}")
print(f"Predictions: {prediction_list}")
```
