## Drawing Hist Plot for all Numerical data

```python
df.hist(figsize=(15, 15))
plt.suptitle("Distribution of Features", fontsize=20)
plt.show()
```

```python
axes = df.hist(figsize=(15, 15), bins=20)

plt.suptitle("Distribution of Features", fontsize=20)

for ax in axes.flatten():
    ax.set_xlabel("Value")
    ax.set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```
