```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/home/tanmoy/Desktop/Phitron/Machine Learning/Module7 - Encoding/Practice/diabetes.csv")
df.sample(5)
```

```python
df.isnull().sum()
```

```python
df["age_group"] = np.select (
    [
        df["Age"] < 30,
        df["Age"].between(30,50, inclusive="both"),
        df["Age"] > 50
    ],
    [
        "young",
        "adult",
        "senior"
    ],
    default="NaN"
)

df.sample(5)
```

```python
df["Age"].isnull().sum()
```

```python
from sklearn.preprocessing import OrdinalEncoder
age_encoded = OrdinalEncoder(categories=[["young", "adult", "senior"]])
age_encoded.fit(df[["age_group"]])
df["age_encoded"] = age_encoded.transform(df[["age_group"]]).ravel()
```
