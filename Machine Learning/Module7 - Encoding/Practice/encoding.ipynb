### Ordinal encoding

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import OrdinalEncoder
```

### dataset load

```python
df = pd.read_csv("/home/tanmoy/Desktop/Phitron/Machine Learning/Module7 - Encoding/Practice/titanic_data_updated.csv")
df.sample(5)
```

## Ordinal Encoding

```python
encoded_pclass = OrdinalEncoder(categories=[['third', 'second', 'first']])
encoded_pclass.fit(df[["Pclass"]])
df["encoded_pclass"] = encoded_pclass.transform(df[["Pclass"]]).ravel()
```

```python
df.sample(5)
```

## One Hot Encoding

```python
from sklearn.preprocessing import OneHotEncoder
sex_one = OneHotEncoder(sparse_output=False).set_output(transform='pandas')
sex_one.fit(df[["Sex"]])
temp = sex_one.transform(df[["Sex"]])

df = pd.concat([df, temp], axis=1)
display(df.sample(5))
```

```python
df["cabin_decki"] = df["Cabin"].astype(str).str[0]
```

```python
df["cabin_decki"].value_counts()
```

```python
cabin_encode = OneHotEncoder(sparse_output=False,drop="first").set_output(transform='pandas')
cabin_encode.fit(df[["cabin_decki"]])
temp = cabin_encode.transform(df[["cabin_decki"]])
df = pd.concat([df, temp], axis=1)
```

```python
df.sample(5)
```
