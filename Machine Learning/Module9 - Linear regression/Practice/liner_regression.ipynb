```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
X = np.array([1,3,4,6,7])
y = np.array([10,30,40,60,70])
```

```python
plt.scatter(X, y)
```

```python
def liner_regression(x, y, w, b):
    m = X.shape[0]
    p_list = np.zeros((m,))
    for i in range(m):
        p_list[i] = (w * x[i]) + b
    return p_list
```

```python
plist = liner_regression(X, y, 0, 0)
print(plist)
```

```python
sns.scatterplot(x = X, y= y)
plt.plot(X, plist)
```

```python
plist = liner_regression(X, y, 5, 0)
print(plist)
sns.scatterplot(x = X, y= y)
plt.plot(X, plist)
```

```python
plist = liner_regression(X, y, 9, 3)
print(plist)
sns.scatterplot(x = X, y= y)
plt.plot(X, plist)
```
