```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
X = np.array([1,3,4,6,7])
y = np.array([10,30,40,60,70])
```

```python
plt.scatter(X, y)
```

```python
def liner_regression(x, y, w, b):
    m = X.shape[0]
    p_list = np.zeros((m,))
    for i in range(m):
        p_list[i] = (w * x[i]) + b
    return p_list
```

```python
plist = liner_regression(X, y, 0, 0)
print(plist)
```

```python
sns.scatterplot(x = X, y= y)
plt.plot(X, plist)
```

```python
plist = liner_regression(X, y, 5, 0)
print(plist)
sns.scatterplot(x = X, y= y)
plt.plot(X, plist)
```

```python
plist = liner_regression(X, y, 9, 3)
print(plist)
sns.scatterplot(x = X, y= y)
plt.plot(X, plist)
```

# Mean Square error calculation

```python
def error_calculation(x, y, w, b):
    m = X.shape[0]
    cost = 0

    for i in range(m):
        prediction = (w * x[i]) + b

        difference = y[i] - prediction

        dif_square = difference **2

        cost = cost + dif_square

    cost = cost / m
    return cost
        
```

```python
plist = liner_regression(X, y, 5, 0)
print(plist)
w = 5
b = 0
sns.scatterplot(x = X, y= y)
plt.plot(X, plist)
plt.title(f"Loss funtion value is {error_calculation(X, y, w, b)} when w = {w} and b = {b}")
```

```python
def error_calculation(x, y, w, b):
    m = x.shape[0]
    cost = 0

    for i in range(m):
        prediction = (w * x[i]) + b

        difference = y[i] - prediction

        dif_square = difference ** 2

        cost = cost + dif_square

    cost = cost / m

    return cost



cost_list = np.zeros(200)
w_values = np.zeros(200)

initial = -100

for i in range(200):
    w_values[i] = initial

    cost_list[i] = error_calculation(X, y, initial, 0)

    initial = initial + 1

plt.plot(w_values, cost_list)

plt.xlabel("w")
plt.ylabel("Cost")
plt.title("Cost Function")

plt.show()
```
